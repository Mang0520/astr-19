```python
# imports the stuff i need to use for this program
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 2*np.pi, 100) # makes my range

# seting my a and b values as said in the problem statmnent 
a = 5
b = 0.5 

# the functiuon for my program
def my_function(x):
    return a * np.exp(-b * x)
```

```python
y = my_function(x) # call my functions for any value of x and get a y value

plt.plot(x, y) # plots the points made in my function

# just labaels on the graph and saving as pdf
plt.xlabel("x")
plt.ylabel("y")

plt.title("y = a*exp(-b*x)")

plt.savefig("plot.pdf")

#shows the graph
plt.show()
```
