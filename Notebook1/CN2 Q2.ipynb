```python
#importing the stuff i need for this program
import numpy as np
import matplotlib.pyplot as plt
```

```python
# making my x values
x = np.linspace(0, 10, 20)

# and than my y valyues
y = 2 * x + 3
```

```python
# uncertainty decreases from 2 to 0.2 using x and making the lines in the graph smaller and smaller
sigma = np.linspace(2, 0.2, len(x))

# adds random scatter
noise = np.random.normal(0, sigma)

# adds the scatter to y
y_scatter = y + noise
```

```python
#add the design to the graph points by using sigma to add the lines, adds color and size 
plt.errorbar(
    x,
    y_scatter,
    yerr=sigma,
    fmt='o',
    color='green',
    ecolor='black',
    capsize=4
)

plt.xlabel("x")
plt.ylabel("y")

plt.savefig("scatter_plot.pdf")

plt.show()
```
