 import math and uses it return sin(x)

```python
import math

def sine_function(x):
    return math.sin(x)
```

and another function that returns the cos(x)

```python
def cosine_function(x):
    return math.cos(x)
```

creates 100 values of x from 0 to 2pi, stores the values of sin,cos, and x in a table. it does this by using a for 

```python
x_values = []

for i in range(1000):
    x = i * (2 * math.pi) / 999
    x_values.append(x)

table = []

for x in x_values:
    table.append([x, sine_function(x), cosine_function(x)])
```

we use a for loop to print out the first 10 values for x, sin, and cos

```python
print(f"{'x':>10} {'sin(x)':>10} {'cos(x)':>10}")

for i in range(10):
    print(f"{table[i][0]:10.4f} {table[i][1]:10.4f} {table[i][2]:10.4f}")
```
